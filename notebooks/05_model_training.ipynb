# 05 Model Training

This notebook creates the saved Random Forest model. For a meaningful model, use a sufficiently large approved project dataset.

```python
import sys
from pathlib import Path
import pandas as pd
sys.path.append(str(Path('..').resolve()))
from src.model import train_model
from src.preprocessing import encode_activity_level

food = pd.read_csv('../data/processed/food_nutrition_clean.csv')
# Demonstration training records derived from nutrition profiles.
# Replace/extend this with the project's real labeled user/profile dataset when available.
rows=[]
for i, r in food.iterrows():
    for activity in ['Sedentary','Light','Moderate','Active','Very Active']:
        for goal in ['Lose Weight','Maintain Weight','Gain Weight']:
            rows.append({
                'age': 18 + (i % 35), 'height_cm': 155 + (i % 30),
                'weight_kg': 45 + (i % 50),
                'activity_level_encoded': encode_activity_level(activity),
                'calories': max(1200, float(r.Calories) * (3 + i % 3)),
                'protein': max(20, float(r.Protein) * (2 + i % 2)),
                'carbohydrates': max(50, float(r.Carbohydrates) * (2 + i % 3)),
                'fat': max(15, float(r.Fat) * (2 + i % 2)),
                'goal_category': goal
            })
train_df = pd.DataFrame(rows)
model, metrics = train_model(train_df, '../models/nutrition_model.joblib')
print(metrics)
print('Saved models/nutrition_model.joblib')
```

**Important:** The generated demonstration labels are not a substitute for a real labeled user dataset. For the final college submission, use actual project training labels and report only metrics calculated from that dataset.
