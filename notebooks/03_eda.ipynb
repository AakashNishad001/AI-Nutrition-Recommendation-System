# 03 Exploratory Data Analysis

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('../data/processed/food_nutrition_clean.csv')
display(df.describe(include='all'))
```

```python
df['Category'].value_counts().plot(kind='bar', figsize=(10,5), title='Food Categories')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
sns.scatterplot(data=df, x='Calories', y='Protein', hue='Diet_Type')
plt.title('Calories vs Protein')
plt.tight_layout()
plt.show()
```

```python
corr = df[['Calories','Protein','Carbohydrates','Fat','Fiber','Sugar','Sodium']].corr()
sns.heatmap(corr, annot=True, fmt='.2f')
plt.title('Nutrient Correlation')
plt.tight_layout()
plt.show()
```
