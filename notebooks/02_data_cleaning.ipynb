```python
import sys
from pathlib import Path
import pandas as pd
sys.path.append(str(Path('..').resolve()))
from src.preprocessing import load_food_data, clean_food_data

raw = Path('../data/raw/food_nutrition_raw.csv')
if not raw.exists():
    raise FileNotFoundError('Put the approved dataset at data/raw/food_nutrition_raw.csv')
df = clean_food_data(load_food_data(raw))
display(df.head())
out = Path('../data/processed/food_nutrition_clean.csv')
out.parent.mkdir(parents=True, exist_ok=True)
df.to_csv(out, index=False)
print('Saved:', out)
```
